### Topic 3: File modes

**Definition**

A file mode tells Python how a file should be opened.

Common modes:
 - r  Read an existing file. Error if the file does not exist.
 - w  Write a file. Creates it or replaces all existing content.
 - a  Append to a file. Creates it or adds content at the end.
 - x  Create a new file. Error if the file already exists.
 - b  Binary mode for non-text data, such as images.
 - t  Text mode. This is the default mode.

**Combined examples**

rb Read a binary file.
wb Write a binary file.

**Important**

Use w carefully because it replaces the file's previous content.

````python
print("\n--- Topic 3: File Modes ---")
print("r = read | w = write | a = append | x = create")



# ------------------------------------------------------------
````

### Topic 5: The with statement

**Definition**

The with statement creates a context in which the file is used.
Python automatically closes the file when that block ends, even if
an error occurs. This is the recommended way to handle files.

**Syntax**

```python

```

### Topic 6: Writing a text file

**Definition**

write() adds a string to an open file. The w mode creates a file or
replaces all its existing content.

write() does not add a new line automatically. Use \n when a line
should end.

writelines() writes multiple strings. Those strings must contain
their own \n characters when separate lines are required.

````python
log_file_path = demo_directory / "pipeline_log.txt"

with open(log_file_path, "w", encoding="utf-8") as log_file:
    log_file.write("Batch 1: Succeeded\n")
    log_file.writelines(
        [
            "Batch 2: Succeeded\n",
            "Batch 3: Failed\n",
        ]
    )

print("\n--- Topic 6: Writing a Text File ---")
print("Log file written:", log_file_path.name)



# ------------------------------------------------------------
````

### Topic 9: Readline(), readlines(), and file iteration

readline(): returns one line at a time.
readlines(): returns all lines as a list of strings.
for line in file: processes one line at a time without loading the
complete file into memory. This is better for large text files.

````python
print("\n--- Topic 9A: readline() ---")
with open(log_file_path, "r", encoding="utf-8") as log_file:
    first_line = log_file.readline().strip()
    second_line = log_file.readline().strip()

print("First line:", first_line)
print("Second line:", second_line)


print("\n--- Topic 9B: readlines() ---")
with open(log_file_path, "r", encoding="utf-8") as log_file:
    all_lines = log_file.readlines()

print("Number of lines:", len(all_lines))
print("Third line:", all_lines[2].strip())


print("\n--- Topic 9C: File Iteration ---")
failed_count = 0

with open(log_file_path, "r", encoding="utf-8") as log_file:
    for line in log_file:
        clean_line = line.strip()

        if "Failed" in clean_line:
            failed_count = failed_count + 1
            print("Failure found:", clean_line)

print("Failed batches:", failed_count)



# ------------------------------------------------------------
````

### Topic 11: Csv dictwriter and dictreader

**Definition**

DictWriter writes dictionaries as CSV rows.
DictReader reads each CSV row as a dictionary whose keys come from
the header. Named columns are usually easier to understand than
numeric indexes.

````python
employees_csv_path = demo_directory / "employees.csv"
employee_columns = ["employee_id", "name", "department"]
employee_records = [
    {"employee_id": 101, "name": "Ananya", "department": "Data"},
    {"employee_id": 102, "name": "Ravi", "department": "Cloud"},
]

with open(employees_csv_path, "w", encoding="utf-8", newline="") as csv_file:
    dictionary_writer = csv.DictWriter(csv_file, fieldnames=employee_columns)
    dictionary_writer.writeheader()
    dictionary_writer.writerows(employee_records)

print("\n--- Topic 11: DictWriter and DictReader ---")
with open(employees_csv_path, "r", encoding="utf-8", newline="") as csv_file:
    dictionary_reader = csv.DictReader(csv_file)

    for employee in dictionary_reader:
        print(
            f"ID: {employee['employee_id']} | "
            f"Name: {employee['name']} | "
            f"Department: {employee['department']}"
        )



# ------------------------------------------------------------
````

### Topic 15: Try and except

**Definition**

Place code that may fail inside try. If the specified exception
occurs, Python runs the matching except block instead of stopping.

**Syntax**

```python
try:
    risky_code
except SpecificError:
    recovery_code
```

Catch specific exception types. Avoid a bare except because it can
hide programming mistakes and make debugging difficult.

````python
print("\n--- Topic 15: try and except ---")
try:
    with open(missing_file_path, "r", encoding="utf-8") as missing_file:
        missing_content = missing_file.read()
except FileNotFoundError:
    print("File could not be read because it does not exist.")



# ------------------------------------------------------------
````

### Topic 16: Accessing the exception message

**Definition**

Use "as" to store the exception object. str(exception) provides the
technical error message, which is useful for application logs.

**Syntax**

```python
except SpecificError as error:
    print(error)
```

````python
print("\n--- Topic 16: Accessing the Exception Message ---")
invalid_number = "one hundred"

try:
    converted_number = int(invalid_number)
except ValueError as conversion_error:
    print("Conversion failed for value:", invalid_number)
    print("Exception type:", type(conversion_error).__name__)



# ------------------------------------------------------------
````

### Topic 17: Handling multiple exception types

**Definition**

One try block may have multiple except blocks. Python runs the first
block matching the exception that occurred.

````python
def read_record_count(file_path: Path) -> int:
    """Read an integer record count from a text file."""
    try:
        with open(file_path, "r", encoding="utf-8") as count_file:
            return int(count_file.read().strip())
    except FileNotFoundError:
        print("Record-count file was not found.")
        return 0
    except ValueError:
        print("Record-count file does not contain a valid integer.")
        return 0
    except PermissionError:
        print("Permission was denied while reading the file.")
        return 0


valid_count_path = demo_directory / "record_count.txt"
invalid_count_path = demo_directory / "invalid_record_count.txt"

with open(valid_count_path, "w", encoding="utf-8") as count_file:
    count_file.write("1500")

with open(invalid_count_path, "w", encoding="utf-8") as count_file:
    count_file.write("fifteen hundred")

print("\n--- Topic 17: Multiple Exception Types ---")
print("Valid count:", read_record_count(valid_count_path))
print("Invalid count:", read_record_count(invalid_count_path))
print("Missing count:", read_record_count(missing_file_path))



# ------------------------------------------------------------
````

### Topic 18: Try, except, else, and finally

**Definition**

try: contains code that may fail.
except: runs when a matching exception occurs.
else: runs only when the try block succeeds.
finally: always runs, whether an exception occurs or not.

finally is commonly used for cleanup. The with statement already
closes files automatically, so it is still preferred for file access.

````python
def load_text_file(file_path: Path) -> str:
    """Load a text file and demonstrate complete exception handling."""
    try:
        with open(file_path, "r", encoding="utf-8") as source_file:
            content = source_file.read()
    except FileNotFoundError:
        print("Load failed: file not found.")
        return ""
    else:
        print("Load succeeded.")
        return content
    finally:
        print("Load attempt completed.")


print("\n--- Topic 18A: Successful try-else-finally Flow ---")
loaded_text = load_text_file(valid_count_path)
print("Loaded value:", loaded_text)


print("\n--- Topic 18B: Failed try-except-finally Flow ---")
loaded_text = load_text_file(missing_file_path)
print("Loaded value:", loaded_text)



# ------------------------------------------------------------
````

### Topic 19: Raising an exception

**Definition**

The raise statement deliberately creates an exception when a value
violates a business rule.

**Syntax**

```python
if invalid_condition:
    raise ValueError("Helpful error message")
```

Raising an error separates validation logic from error handling.

````python
def calculate_rejection_percentage(
    total_records: int,
    rejected_records: int,
) -> float:
    """Return rejection percentage after validating record counts."""
    if total_records <= 0:
        raise ValueError("Total records must be greater than zero.")

    if rejected_records < 0:
        raise ValueError("Rejected records cannot be negative.")

    if rejected_records > total_records:
        raise ValueError("Rejected records cannot exceed total records.")

    return rejected_records / total_records * 100


print("\n--- Topic 19: Raising an Exception ---")
try:
    rejection_percentage = calculate_rejection_percentage(1000, 25)
    print("Rejection percentage:", rejection_percentage)
except ValueError as validation_error:
    print("Validation failed:", validation_error)

try:
    rejection_percentage = calculate_rejection_percentage(0, 0)
    print("Rejection percentage:", rejection_percentage)
except ValueError as validation_error:
    print("Validation failed:", validation_error)



# ------------------------------------------------------------
````

### Topic 20: Reliable csv processing example

**Scenario**

Read a sales CSV file, convert each amount to a number, skip invalid
values, and return a processing summary.

````python
def process_sales_file(file_path: Path) -> dict:
    """Process a sales CSV file and return record-count statistics."""
    total_records = 0
    valid_records = 0
    rejected_records = 0
    total_sales = 0.0

    try:
        with open(file_path, "r", encoding="utf-8", newline="") as csv_file:
            sales_reader = csv.DictReader(csv_file)

            required_columns = {"sale_id", "product", "amount"}
            actual_columns = set(sales_reader.fieldnames or [])

            if not required_columns.issubset(actual_columns):
                raise ValueError("CSV file is missing required columns.")

            for sales_record in sales_reader:
                total_records = total_records + 1

                try:
                    amount = float(sales_record["amount"])

                    if amount <= 0:
                        raise ValueError("Amount must be positive.")
                except (TypeError, ValueError):
                    rejected_records = rejected_records + 1
                    continue

                valid_records = valid_records + 1
                total_sales = total_sales + amount

    except FileNotFoundError:
        return {
            "status": "Failed: file not found",
            "total_records": 0,
            "valid_records": 0,
            "rejected_records": 0,
            "total_sales": 0.0,
        }
    except (PermissionError, ValueError) as processing_error:
        return {
            "status": f"Failed: {processing_error}",
            "total_records": total_records,
            "valid_records": valid_records,
            "rejected_records": rejected_records,
            "total_sales": total_sales,
        }
    else:
        return {
            "status": "Succeeded",
            "total_records": total_records,
            "valid_records": valid_records,
            "rejected_records": rejected_records,
            "total_sales": total_sales,
        }


quality_csv_path = demo_directory / "sales_quality_example.csv"
quality_rows = [
    {"sale_id": 1, "product": "Laptop", "amount": 65000},
    {"sale_id": 2, "product": "Mouse", "amount": "invalid"},
    {"sale_id": 3, "product": "Keyboard", "amount": 2500},
    {"sale_id": 4, "product": "Monitor", "amount": -500},
]

with open(quality_csv_path, "w", encoding="utf-8", newline="") as csv_file:
    quality_writer = csv.DictWriter(
        csv_file,
        fieldnames=["sale_id", "product", "amount"],
    )
    quality_writer.writeheader()
    quality_writer.writerows(quality_rows)

print("\n--- Topic 20: Reliable CSV Processing Example ---")
processing_summary = process_sales_file(quality_csv_path)

for summary_key, summary_value in processing_summary.items():
    print(f"{summary_key}: {summary_value}")



# ============================================================
````

## 4. KEY TAKEAWAYS

1. A file preserves data after a Python program ends.
