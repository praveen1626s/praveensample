### Topic 1: Introduction to Collections

**Definition**

A collection is a single variable that stores multiple values. Collections make it easier to organize and process related data.

**Common built-in collections**

| Collection | Ordered | Changeable | Duplicates | Main use |
|---|---:|---:|---:|---|
| List | Yes | Yes | Yes | Ordered items that may change |
| Tuple | Yes | No | Yes | Fixed ordered values |
| Set | No | Yes | No | Unique values |
| Dictionary | Yes | Yes | Keys: No | Key-value records |

````python
source_files = ["customers.csv", "products.csv", "sales.csv"]

print("Source files:", source_files)
print("Number of files:", len(source_files))
````

### Topic 2: Lists

**Definition**

A list is an ordered and changeable collection. It can contain values of the same type or different types, and duplicate values are allowed.

**Syntax**

```python
list_name = [item_1, item_2, item_3]
```

**Characteristics**

- Ordered
- Mutable
- Indexed
- Allows duplicates

````python
aws_services = ["S3", "Glue", "Athena", "S3"]
batch_details = [101, "sales.csv", 2500, True]

print("AWS services:", aws_services)
print("Batch details:", batch_details)
print("Collection type:", type(aws_services))
````

### Topic 3: List Indexing

**Definition**

Indexing accesses one item using its position. Python uses zero-based indexing, so the first item is at index `0`.

Negative indexing starts from the end:

| Value | S3 | Glue | Athena | Redshift |
|---|---:|---:|---:|---:|
| Positive index | 0 | 1 | 2 | 3 |
| Negative index | -4 | -3 | -2 | -1 |

**Syntax**

```python
list_name[index]
```

Accessing a position that does not exist raises `IndexError`.

````python
pipeline_services = ["S3", "Glue", "Athena", "Redshift"]

print("First service:", pipeline_services[0])
print("Third service:", pipeline_services[2])
print("Last service:", pipeline_services[-1])
````

### Topic 4: List Slicing

**Definition**

Slicing extracts a section of a list and returns a new list. The start index is included, while the stop index is excluded.

**Syntax**

```python
list_name[start:stop]
list_name[start:stop:step]
```

````python
data_layers = ["Landing", "Bronze", "Silver", "Gold", "Archive"]

print("Indexes 1 to 3:", data_layers[1:4])
print("First three:", data_layers[:3])
print("From index 2:", data_layers[2:])
print("Every second item:", data_layers[::2])
print("Reversed list:", data_layers[::-1])
````

### Topic 5: Adding List Items

Lists are mutable, so their contents can change after creation.

| Method | Purpose |
|---|---|
| `append(item)` | Adds one item to the end |
| `insert(index, item)` | Adds one item at a specific position |
| `extend(iterable)` | Adds all items from another collection |

````python
file_queue = ["customers.csv", "products.csv"]

file_queue.append("sales.csv")
print("After append:", file_queue)

file_queue.insert(1, "countries.csv")
print("After insert:", file_queue)

file_queue.extend(["returns.csv", "stores.csv"])
print("After extend:", file_queue)
````

### Topic 6: Updating and Removing List Items

**Updating syntax**

```python
list_name[index] = new_value
```

**Removal options**

| Operation | Purpose |
|---|---|
| `remove(value)` | Removes the first matching value |
| `pop(index)` | Removes and returns the item at an index |
| `pop()` | Removes and returns the last item |
| `del list_name[index]` | Deletes an item at an index |
| `clear()` | Removes every item |

````python
job_statuses = ["Pending", "Running", "Failed", "Waiting"]

job_statuses[2] = "Succeeded"
print("After update:", job_statuses)

job_statuses.remove("Waiting")
print("After remove:", job_statuses)

removed_status = job_statuses.pop(0)
print("Popped value:", removed_status)
print("After pop:", job_statuses)

del job_statuses[0]
print("After del:", job_statuses)
````

### Topic 7: Useful List Operations and Methods

| Operation | Purpose |
|---|---|
| `len(list)` | Returns the number of items |
| `value in list` | Checks whether a value exists |
| `count(value)` | Counts matching values |
| `index(value)` | Returns the first matching index |
| `sort()` | Sorts the original list |
| `reverse()` | Reverses the original list |
| `min()` | Returns the smallest numeric value |
| `max()` | Returns the largest numeric value |
| `sum()` | Returns the numeric total |

````python
record_counts = [450, 120, 900, 120, 600]

print("Number of batches:", len(record_counts))
print("Contains 900:", 900 in record_counts)
print("Count of 120:", record_counts.count(120))
print("Index of 900:", record_counts.index(900))
print("Minimum count:", min(record_counts))
print("Maximum count:", max(record_counts))
print("Total records:", sum(record_counts))

record_counts.sort()
print("Sorted counts:", record_counts)

record_counts.reverse()
print("Descending counts:", record_counts)
````

### Topic 8: Copying a List

**Definition**

Assigning one list variable to another does not create an independent list. Both names refer to the same list.

Use `copy()` when an independent shallow copy is required.

**Syntax**

```python
copied_list = original_list.copy()
```

````python
original_files = ["a.csv", "b.csv"]
copied_files = original_files.copy()

copied_files.append("c.csv")

print("Original:", original_files)
print("Copy:", copied_files)
````

### Topic 9: Tuples

**Definition**

A tuple is an ordered collection that cannot be changed after it is created. This property is called immutability.

**Syntax**

```python
tuple_name = (item_1, item_2, item_3)
```

**Characteristics**

- Ordered and indexed
- Allows duplicates
- Cannot add, replace, or remove individual items
- Useful for fixed values

A one-item tuple requires a trailing comma:

```python
one_item_tuple = ("S3",)
```

````python
database_connection = ("analytics-db", 5432, "production")

print("Connection:", database_connection)
print("Host:", database_connection[0])
print("Port:", database_connection[1])
print("Tuple length:", len(database_connection))
print("Collection type:", type(database_connection))
````

### Topic 10: Tuple Packing and Unpacking

**Definition**

Packing places multiple values into one tuple. Unpacking assigns those values to separate variables.

**Syntax**

```python
packed_tuple = value_1, value_2
variable_1, variable_2 = packed_tuple
```

The number of variables should match the number of unpacked values.

````python
batch_info = (501, "sales.csv", "Succeeded")
batch_id, file_name, batch_status = batch_info

print("Batch ID:", batch_id)
print("File name:", file_name)
print("Status:", batch_status)
````

### Topic 11: Sets

**Definition**

A set is an unordered collection of unique values. Duplicate values are automatically removed.

**Syntax**

```python
set_name = {item_1, item_2, item_3}
empty_set = set()
```

**Important**

- `{}` creates an empty dictionary, not an empty set.
- Sets do not support indexing or slicing.
- Display order may vary between runs.

````python
raw_regions = {"South", "North", "South", "West", "North"}

print("Unique regions:", raw_regions)
print("Number of unique regions:", len(raw_regions))
print("Contains South:", "South" in raw_regions)
````

### Topic 12: Adding and Removing Set Items

| Method | Purpose |
|---|---|
| `add(item)` | Adds one item |
| `update(iterable)` | Adds multiple items |
| `remove(item)` | Removes an item and raises `KeyError` if missing |
| `discard(item)` | Removes an item without an error if missing |
| `clear()` | Removes all items |

````python
file_formats = {"CSV", "JSON"}

file_formats.add("PARQUET")
file_formats.update({"AVRO", "ORC"})
print("After additions:", file_formats)

file_formats.remove("JSON")
file_formats.discard("XML")
print("After removals:", file_formats)
````

### Topic 13: Set Operations

| Operation | Operator | Purpose |
|---|---|---|
| Union | `|` | Values found in either set |
| Intersection | `&` | Values common to both sets |
| Difference | `-` | Values in the first set but not the second |
| Symmetric difference | `^` | Values found in exactly one set |

````python
source_columns = {"id", "name", "email", "city"}
target_columns = {"id", "name", "email", "country"}

print("All columns:", source_columns | target_columns)
print("Common columns:", source_columns & target_columns)
print("Only in source:", source_columns - target_columns)
print("Mismatched columns:", source_columns ^ target_columns)
````

### Topic 14: Dictionaries

**Definition**

A dictionary stores data as key-value pairs. Every key is unique and is used to access its associated value.

**Syntax**

```python
dictionary_name = {
    "key_1": value_1,
    "key_2": value_2
}
```

**Characteristics**

- Maintains insertion order in modern Python
- Mutable
- Keys must be unique
- Values may be duplicated

````python
pipeline = {
    "pipeline_id": 1001,
    "name": "daily_sales_load",
    "source": "Amazon S3",
    "record_count": 5000,
    "completed": True,
}

print("Pipeline:", pipeline)
print("Pipeline name:", pipeline["name"])
print("Source:", pipeline.get("source"))
print("Collection type:", type(pipeline))
````

### Topic 15: Accessing Dictionary Values

**Bracket notation**

```python
dictionary[key]
```

It returns the value but raises `KeyError` when the key does not exist.

**The `get()` method**

```python
dictionary.get(key)
dictionary.get(key, default)
```

It returns `None` or a chosen default when the key is missing.

````python
job = {"job_id": 701, "status": "Running"}

print("Job ID:", job["job_id"])
print("Status:", job.get("status"))
print("Owner:", job.get("owner"))
print("Priority:", job.get("priority", "Normal"))
````

### Topic 16: Adding, Updating, and Removing Dictionary Items

**Add or update**

```python
dictionary[key] = value
dictionary.update({key: value})
```

**Remove**

| Operation | Purpose |
|---|---|
| `pop(key)` | Removes a key and returns its value |
| `del dictionary[key]` | Deletes a key-value pair |
| `clear()` | Removes all pairs |

````python
batch = {"batch_id": 301, "status": "Pending"}

batch["source"] = "sales.csv"
print("After addition:", batch)

batch["status"] = "Succeeded"
batch.update({"record_count": 1200})
print("After updates:", batch)

removed_source = batch.pop("source")
print("Removed source:", removed_source)

del batch["record_count"]
print("Final batch:", batch)
````

### Topic 17: Dictionary Methods

| Method | Returns |
|---|---|
| `keys()` | All keys |
| `values()` | All values |
| `items()` | All key-value pairs |

````python
dataset = {
    "name": "customers",
    "format": "Parquet",
    "rows": 8500,
}

print("Keys:", list(dataset.keys()))
print("Values:", list(dataset.values()))
print("Items:", list(dataset.items()))
````

### Topic 18: Iterating Through Collections

A `for` loop can process every item in a collection.

**List syntax**

```python
for item in list_name:
    code
```

**Dictionary syntax**

```python
for key, value in dictionary_name.items():
    code
```

````python
files_to_process = ["customers.csv", "products.csv", "sales.csv"]

for current_file in files_to_process:
    print("Processing:", current_file)

print("---")

table_counts = {
    "customers": 500,
    "products": 200,
    "sales": 1500,
}

for table_name, row_count in table_counts.items():
    print(f"{table_name}: {row_count} rows")
````

### Topic 19: Nested Collections

**Definition**

A nested collection is a collection stored inside another collection.

A list of dictionaries is commonly used to represent multiple records:

- Each dictionary represents one record.
- The list stores all records.
- Dictionary keys represent field names.

````python
sales_records = [
    {"sale_id": 1, "product": "Laptop", "amount": 65000},
    {"sale_id": 2, "product": "Mouse", "amount": 1200},
    {"sale_id": 3, "product": "Keyboard", "amount": 2500},
]

print("First record:", sales_records[0])
print("First product:", sales_records[0]["product"])

total_sales = 0

for sales_record in sales_records:
    print(
        f"Sale {sales_record['sale_id']}: "
        f"{sales_record['product']} - {sales_record['amount']}"
    )
    total_sales = total_sales + sales_record["amount"]

print("Total sales amount:", total_sales)
````

### Topic 20: Choosing the Correct Collection

**Use a list when:**

- Order matters.
- Values may change.
- Duplicates are acceptable.
- Example: a queue of files to process.

**Use a tuple when:**

- Order matters.
- Values should remain fixed.
- Example: database connection information.

**Use a set when:**

- Only unique values are needed.
- Membership checks or set comparisons are important.
- Example: unique country codes or schema comparison.

**Use a dictionary when:**

- Every value needs a meaningful key.
- One record contains several named attributes.
- Example: a pipeline configuration or customer record.

````python
print("List: ordered and changeable")
print("Tuple: ordered and fixed")
print("Set: unique values")
print("Dictionary: key-value pairs")
````

## 4. KEY TAKEAWAYS

1. Collections store multiple related values in one variable.
2. Lists are ordered, mutable, indexed, and allow duplicates.
3. Indexing retrieves one item; slicing retrieves a range of items.
4. Tuples are ordered but cannot be changed after creation.
5. Sets store unique values and support mathematical set operations.
6. Dictionaries store unique keys mapped to values.
7. `get()` safely reads a dictionary key that may not exist.
8. `for` loops can process items, keys, values, or key-value pairs.
9. A list of dictionaries is useful for representing tabular records.
10. Select a collection based on order, mutability, uniqueness, and access requirements.
