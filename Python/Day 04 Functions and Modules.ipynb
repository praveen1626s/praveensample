### Topic 1: Introduction to functions

**Definition**

A function is a named, reusable block of code that performs a
specific task. It runs only when it is called.

Functions help us:

- Avoid repeating the same code.
- Divide a large program into smaller tasks.
- Improve readability, testing, and maintenance.
- Reuse transformation and validation logic.

Example without a function:

````python
print("\n--- Topic 1: Introduction to Functions ---")
print("Validating customers.csv")
print("Validating products.csv")

# The word "Validating" is repeated. A function can make this logic
# reusable, as shown in the next topic.
#


# ------------------------------------------------------------
````

### Topic 2: Defining and calling a function

**Definition**

Use the def keyword to define a function. Calling the function by
name executes the code inside it.

**Syntax**

```python
def function_name():
    statements
```

function_name()

**Important**

- Add parentheses after the function name.
- Add a colon after the function header.
- Indent every statement inside the function.
- Use snake_case for function names.

````python
def display_pipeline_message():
    """Display a simple pipeline message."""
    print("The data pipeline has started.")


print("\n--- Topic 2: Defining and Calling a Function ---")
display_pipeline_message()
display_pipeline_message()



# ------------------------------------------------------------
````

### Topic 3: Parameters and arguments

**Definition**

A parameter is a variable listed in a function definition.
An argument is the actual value supplied when calling the function.

**Syntax**

```python
def function_name(parameter):
    statements
```

function_name(argument)

In this example, file_name is a parameter and "sales.csv" is an
argument.

````python
def display_file_status(file_name):
    """Display the processing status of one file."""
    print(f"Processing file: {file_name}")


print("\n--- Topic 3: Parameters and Arguments ---")
display_file_status("customers.csv")
display_file_status("sales.csv")



# ------------------------------------------------------------
````

### Topic 4: Returning a value

**Definition**

The return statement sends a result back to the code that called
the function. The returned result can be stored or used elsewhere.

**Syntax**

```python
def function_name(parameters):
    result = calculation
    return result
```

returned_value = function_name(arguments)

**Important**

print() displays a value, while return gives the value back to the
caller. Code after an executed return statement does not run.

````python
def calculate_loaded_records(received_records, rejected_records):
    """Return the number of successfully loaded records."""
    loaded_records = received_records - rejected_records
    return loaded_records


print("\n--- Topic 4: Returning a Value ---")
successful_records = calculate_loaded_records(1000, 25)
print("Successfully loaded records:", successful_records)



# ------------------------------------------------------------
````

### Topic 5: Multiple parameters

**Definition**

A function can accept multiple parameters separated by commas.
Positional arguments are matched to parameters by their order.

**Example**

````python
def calculate_total_amount(unit_price, quantity, discount):
    """Calculate the final amount after subtracting a discount."""
    gross_amount = unit_price * quantity
    final_amount = gross_amount - discount
    return final_amount


print("\n--- Topic 5: Multiple Parameters ---")
invoice_total = calculate_total_amount(2500.0, 4, 500.0)
print("Invoice total:", invoice_total)



# ------------------------------------------------------------
````

### Topic 6: Default arguments

**Definition**

A default parameter already has a value in the function definition.
The caller may omit its argument or provide a different value.

**Syntax**

```python
def function_name(required_parameter, optional_parameter=default):
    statements
```

**Important**

Parameters without default values must come before parameters with
default values.

````python
def display_job_status(job_name, status="Pending"):
    """Display a job name and its status."""
    print(f"Job: {job_name} | Status: {status}")


print("\n--- Topic 6: Default Arguments ---")
display_job_status("daily_customer_load")
display_job_status("daily_sales_load", "Succeeded")



# ------------------------------------------------------------
````

### Topic 7: Keyword arguments

**Definition**

A keyword argument identifies its parameter by name. This improves
clarity and allows arguments to be passed in a different order.

**Syntax**

```python
function_name(parameter_2=value_2, parameter_1=value_1)
```

````python
def display_batch(batch_id, source_name, record_count):
    """Display a formatted batch summary."""
    print(
        f"Batch: {batch_id} | Source: {source_name} | "
        f"Records: {record_count}"
    )


print("\n--- Topic 7: Keyword Arguments ---")
display_batch(
    record_count=5000,
    batch_id=301,
    source_name="Amazon S3",
)



# ------------------------------------------------------------
````

### Topic 8: Returning multiple values

**Definition**

A function can return multiple comma-separated values. Python packs
them into a tuple, which can be unpacked into separate variables.

**Syntax**

```python
def function_name():
    return value_1, value_2
```

result_1, result_2 = function_name()

````python
def calculate_record_summary(total_records, rejected_records):
    """Return accepted records and rejection percentage."""
    accepted_records = total_records - rejected_records
    rejection_percentage = rejected_records / total_records * 100
    return accepted_records, rejection_percentage


print("\n--- Topic 8: Returning Multiple Values ---")
accepted, rejection_rate = calculate_record_summary(2000, 50)
print("Accepted records:", accepted)
print("Rejection percentage:", rejection_rate)



# ------------------------------------------------------------
````

### Topic 9: Variable scope

**Definition**

Scope describes where a variable can be accessed.

Local variable:
Created inside a function and normally available only there.

Global variable:
Created outside functions and available throughout the module.

**Recommendation**

Prefer parameters and return values instead of modifying global
variables inside functions. This keeps functions predictable.

````python
environment = "Development"  # Global variable


def show_scope():
    """Demonstrate global and local variable access."""
    pipeline_name = "customer_pipeline"  # Local variable
    print("Inside function - Environment:", environment)
    print("Inside function - Pipeline:", pipeline_name)


print("\n--- Topic 9: Variable Scope ---")
show_scope()
print("Outside function - Environment:", environment)


# This would cause NameError because pipeline_name is local:
# print(pipeline_name)


# ------------------------------------------------------------
````

### Topic 10: Type hints

**Definition**

Type hints document the expected parameter types and return type.
They improve readability and help code editors detect mistakes.
Python does not automatically enforce type hints at runtime.

**Syntax**

```python
def function_name(parameter: type) -> return_type:
    statements
```

````python
def calculate_file_size_mb(size_in_bytes: int) -> float:
    """Convert a file size from bytes to megabytes."""
    size_in_mb = size_in_bytes / (1024 * 1024)
    return size_in_mb


print("\n--- Topic 10: Type Hints ---")
file_size_mb = calculate_file_size_mb(10_485_760)
print("File size in MB:", file_size_mb)



# ------------------------------------------------------------
````

### Topic 11: Docstrings

**Definition**

A docstring explains what a function, class, or module does. It is
written as the first triple-quoted string inside the definition.

**Syntax**

```python
def function_name():
    """Short description of the function."""
```

The built-in help() function can display a docstring. The special
__doc__ attribute can also access it directly.

````python
def validate_record_count(actual_count: int, expected_count: int) -> bool:
    """Return True when actual and expected record counts match."""
    return actual_count == expected_count


print("\n--- Topic 11: Docstrings ---")
print(validate_record_count.__doc__)
print("Counts match:", validate_record_count(500, 500))



# ------------------------------------------------------------
````

### Topic 12: Lambda expressions

**Definition**

A lambda is a small anonymous function written in one expression.
It is useful for short, simple operations.

**Syntax**

```python
lambda parameters: expression
```

Normal functions are usually clearer for reusable or complex logic.
Lambda expressions cannot contain multiple statements.

**Example**

````python
calculate_tax = lambda amount, tax_rate: amount * tax_rate / 100

print("\n--- Topic 12: Lambda Expressions ---")
tax_amount = calculate_tax(5000, 18)
print("Tax amount:", tax_amount)



# ------------------------------------------------------------
````

### Topic 13: Using a lambda with sorted()

**Definition**

The key parameter of sorted() accepts a function that decides which
value should be used for sorting. A lambda is useful for this.

sorted() returns a new list and does not change the original list.

**Example**

````python
pipeline_runs = [
    {"name": "customers", "records": 800},
    {"name": "sales", "records": 2500},
    {"name": "products", "records": 400},
]

sorted_runs = sorted(
    pipeline_runs,
    key=lambda run: run["records"],
    reverse=True,
)

print("\n--- Topic 13: Lambda with sorted() ---")
for pipeline_run in sorted_runs:
    print(pipeline_run["name"], pipeline_run["records"])



# ------------------------------------------------------------
````

### Topic 14: Built-in functions

**Definition**

Built-in functions are provided by Python and can be used without
importing anything.

**Common examples**

len()    Number of items
sum()    Total of numeric values
min()    Smallest value
max()    Largest value
sorted() New sorted list
type()   Data type of a value
round()  Rounded numeric value

**Example**

### Topic 15: Introduction to modules

**Definition**

A module is a Python file containing reusable variables, functions,
or classes. Python's standard library includes many ready-to-use
modules, such as math and datetime.

**Benefits**

- Keeps related code together.
- Reduces duplication.
- Makes code reusable across files.
- Provides tested standard functionality.

**Basic syntax**

```python
import module_name
module_name.function_name()
```

````python
import math

print("\n--- Topic 15: Introduction to Modules ---")
print("Square root of 144:", math.sqrt(144))
print("10 rounded upward:", math.ceil(10.2))
print("10 rounded downward:", math.floor(10.8))



# ------------------------------------------------------------
````

### Topic 16: Different ways to import

Method 1: Import the complete module.
import math
result = math.sqrt(25)

Method 2: Import a specific item.
from math import sqrt
result = sqrt(25)

Method 3: Import with an alias.
import datetime as dt
date_value = dt.date(2026, 9, 8)

**Recommendation**

Use clear imports and avoid "from module import *" because it can
create naming conflicts and makes code harder to understand.

````python
from math import pow, sqrt
import datetime as dt

print("\n--- Topic 16: Different Ways to Import ---")
print("Square root:", sqrt(81))
print("Power:", pow(2, 5))

scheduled_date = dt.date(2026, 9, 8)
print("Scheduled date:", scheduled_date)
print("Year:", scheduled_date.year)



# ------------------------------------------------------------
````

### Topic 18: __name__ and the main guard

**Definition**

Python automatically creates a variable named __name__.

When a file is run directly:
__name__ contains "__main__".

When a file is imported:
__name__ contains the module's name.

The main guard prevents selected code from running automatically
when the file is imported by another module.

**Syntax**

```python
if __name__ == "__main__":
    code_to_run_only_when_executed_directly
```

````python
def display_main_guard_example():
    """Display a message from the main-guard example."""
    print("This file is being executed directly.")


print("\n--- Topic 18: __name__ and Main Guard ---")
print("Current __name__ value:", __name__)

if __name__ == "__main__":
    display_main_guard_example()



# ------------------------------------------------------------
````

### Topic 19: Reusable data-engineering functions

**Scenario**

Validate a batch and calculate its successfully loaded records.
Each function performs one clear responsibility.

````python
def is_valid_batch(total_records: int, rejected_records: int) -> bool:
    """Return True when batch record counts are logically valid."""
    return total_records >= 0 and 0 <= rejected_records <= total_records


def get_load_status(total_records: int, rejected_records: int) -> str:
    """Return a readable load status for a batch."""
    if not is_valid_batch(total_records, rejected_records):
        return "Invalid batch counts"

    if total_records == 0:
        return "Empty batch"

    if rejected_records == 0:
        return "Completed without rejections"

    return "Completed with rejections"


def create_batch_summary(
    batch_id: int,
    total_records: int,
    rejected_records: int,
) -> dict:
    """Create and return a dictionary containing a batch summary."""
    if not is_valid_batch(total_records, rejected_records):
        accepted_records = 0
    else:
        accepted_records = total_records - rejected_records

    return {
        "batch_id": batch_id,
        "total_records": total_records,
        "accepted_records": accepted_records,
        "rejected_records": rejected_records,
        "status": get_load_status(total_records, rejected_records),
    }


print("\n--- Topic 19: Reusable Data-Engineering Functions ---")
batch_summary = create_batch_summary(
    batch_id=501,
    total_records=1000,
    rejected_records=25,
)

for summary_key, summary_value in batch_summary.items():
    print(f"{summary_key}: {summary_value}")



# ============================================================
````

## 4. KEY TAKEAWAYS

1. Functions are named, reusable blocks that perform specific tasks.
2. Parameters appear in definitions; arguments appear in calls.
3. return sends a result back, while print() only displays a value.
4. Default arguments provide optional values; keyword arguments make
   calls clearer and allow a different argument order.

5. Local variables belong to a function; global variables are defined
   outside functions.

6. Type hints document expected types but do not enforce them by
   themselves.

7. Docstrings explain the purpose and behavior of reusable code.
8. Lambdas are suitable for short, single-expression functions.
9. Modules organize reusable Python code in separate files.
10. The main guard separates direct execution from imported behavior.
11. Small, focused functions make data pipelines easier to test,
    reuse, debug, and maintain.
