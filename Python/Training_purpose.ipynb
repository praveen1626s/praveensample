```python
expected_count = 750
actual_count = 748

if actual_count == expected_count:
    print("Validation passed: record counts match.")
else:
    print(f"Validation failed: actual {actual_count}, expected {expected_count},missing {expected_count-actual_count}.")
```

```python
def validate_count(expected_count, actual_count):
    if expected_count == actual_count:
        return "Validation Passed"
    else:
        return "Validation Failed"

result = validate_count(750, 750)

print(result)
```
