```python
# DAY 01: PYTHON FUNDAMENTALS
# 1. TOPICS COVERED
# 1. Variables
# 2. Python naming rules
# 3. Basic data types: int, float, str, bool, and None
# 4. Checking data types with type()
# 5. Arithmetic, comparison, and logical operators
# 6. Type conversion
# 7. Displaying output with print()
# 8. Reading user input with input()
# 2. TOPIC EXPLANATIONS
# Topic 1: VARIABLES
#
# Definition:
# A variable is a name that refers to a value stored in memory.
# Variables help us reuse and update data without repeating values.
#
# Syntax:
# variable_name = value
#
# Important:
# The equals sign (=) assigns the value on its right to the name
# on its left. It does not mean "is equal to" in this context.
#
# Example:
```

```python
student_name = "Ananya"
student_age = 22
course_name = "Python for "

print("\n--- Topic 1: Variables ---")
print("Student name:", student_name)
print("Student age:", student_age)
print("Course:", course_name)
```

```python
# Topic 2: PYTHON NAMING RULES
#
# Definition:
# A variable name should clearly describe the value it stores.
#
# Rules:
# - A name may contain letters, numbers, and underscores.
# - A name cannot begin with a number.
# - A name cannot contain spaces or symbols such as @, #, or -.
# - Python names are case-sensitive: age and Age are different.
# - Python keywords such as if, for, and class cannot be names.
# - Use snake_case for normal Python variable names.
#
# Valid names:
```

```python
employee_id = 1001
monthly_salary = 45000.00
is_active = True
```

```python
# Invalid names are comments so that this file remains runnable:
# 1employee = "Arun"       # Cannot start with a number
# employee-name = "Arun"   # Hyphen is not allowed
# class = "Python"         # class is a Python keyword
```

```python
print("\n--- Topic 2: Naming Rules ---")
print("Employee ID:", employee_id)
print("Monthly salary:", monthly_salary)
print("Is active:", is_active)
```

```python
# Topic 3: BASIC DATA TYPES
#
# Definition:
# A data type tells Python what kind of value is being stored and
# which operations can be performed on that value.
#
# Common types:
# int   : Whole numbers, such as 10, 0, and -25
# float : Decimal numbers, such as 10.5 and -2.75
# str   : Text enclosed in quotes, such as "Python"
# bool  : Either True or False
# None  : Represents the absence of a value
#
# Syntax and examples:
```

```python
record_count = 1500                 # int
average_amount = 875.50             # float
source_system = "Amazon S3"         # str
load_completed = True               # bool
error_message = None                # NoneType

print("\n--- Topic 3: Basic Data Types ---")
print("Record count:", record_count)
print("Average amount:", average_amount)
print("Source system:", source_system)
print("Load completed:", load_completed)
print("Error message:", error_message)
```

```python
# Topic 4: CHECKING DATA TYPES
#
# Definition:
# The built-in type() function returns the data type of a value.
# This is useful when inspecting data received from files, APIs,
# databases, or users.
#
# Syntax:
# type(value_or_variable)
#
# Example:
```

```python
print("\n--- Topic 4: Checking Data Types ---")
print(type(record_count))
print(type(average_amount))
print(type(source_system))
print(type(load_completed))
print(type(error_message))
```

```python
# TOPIC 5: OPERATORS
# - 
# Definition:
# Operators are symbols or words that perform operations on values.
# A. Arithmetic operators
# +   Addition
# - Subtraction
# *   Multiplication
# /   Division (returns a float)
# //  Floor division (removes the decimal part)
# %   Modulus (returns the remainder)
# **  Exponentiation (raises a number to a power)
```

```python
total_records = 105
batch_size = 20

print("\n--- Topic 5A: Arithmetic Operators ---")
print("Addition:", 10 + 5)
print("Subtraction:", 10 - 5)
print("Multiplication:", 10 * 5)
print("Division:", 10 / 5)
print("Floor division:", total_records // batch_size)
print("Remainder:", total_records % batch_size)
print("Power:", 2 ** 3)
```

```python
# B. Comparison operators
# ==  Equal to
# !=  Not equal to
# Greater than
# <   Less than
# =  Greater than or equal to
# <=  Less than or equal to
# A comparison always produces True or False.
```

```python
processed_records = 100
expected_records = 100

print("\n--- Topic 5B: Comparison Operators ---")
print("Counts match:", processed_records == expected_records)
print("Counts differ:", processed_records != expected_records)
print("Processed more than 50:", processed_records > 50)
print("Processed at most 100:", processed_records <= 100)
```

```python
# Expected output:
# Counts match: True
# Counts differ: False
# Processed more than 50: True
# Processed at most 100: True
# C. Logical operators
# and : True only when both conditions are True
# or  : True when at least one condition is True
# not : Reverses True to False, or False to True
```

```python
file_exists = True
schema_is_valid = True
has_errors = False

print("\n--- Topic 5C: Logical Operators ---")
print("Ready to process:", file_exists and schema_is_valid)
print("Needs attention:", has_errors or not schema_is_valid)
print("No errors:", not has_errors)
```

```python
# TOPIC 6: TYPE CONVERSION
# - 
# Definition:
# Type conversion changes a value from one data type to another.
# It is also called type casting.
# Common conversion functions:
# int(value)   -> converts a compatible value to an integer
# float(value) -> converts a compatible value to a decimal number
# str(value)   -> converts a value to text
# bool(value)  -> converts a value to True or False
# Important:
# The value must be compatible with the requested type.
# int("25") is valid, but int("twenty-five") causes an error.
```

```python
records_as_text = "250"
price_as_text = "99.50"

records_as_number = int(records_as_text)
price_as_number = float(price_as_text)
job_id_as_text = str(5001)

print("\n--- Topic 6: Type Conversion ---")
print(records_as_number, type(records_as_number))
print(price_as_number, type(price_as_number))
print(job_id_as_text, type(job_id_as_text))
```

```python
# TOPIC 7: DISPLAYING OUTPUT WITH print()
# - 
# Definition:
# The print() function displays text and values on the screen.
# Syntax:
# print(value1, value2, ...)
# An f-string places variable values directly inside a string.
# Syntax: f"Text {variable_name}"
```

```python
pipeline_name = "daily-sales-load"
rows_loaded = 1250

print("\n--- Topic 7: Displaying Output ---")
print("Pipeline:", pipeline_name)
print(f"The {pipeline_name} pipeline loaded {rows_loaded} rows.")
```
